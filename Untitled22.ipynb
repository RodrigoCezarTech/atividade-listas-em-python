```python
# PARTE 1 - Criar listas de produtos e preços

produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

print("Produtos cadastrados:")

for i, j in zip(produtos, precos):
    print(f"{i} - R$ {j:.2f}")
```

```python
# PARTE 2 - Cadastrar novo produto

nome = input("\nDigite o nome do novo produto: ")
preco = float(input("Digite o preço do produto: "))

produtos.append(nome)
precos.append(preco)

print("\nEstoque atualizado:")

for i, j in zip(produtos, precos):
    print(f"{i} - R$ {j:.2f}")
```

```python
# PARTE 3 - Remover o último produto

if len(produtos) > 0:
    removido = produtos.pop()
    preco_removido = precos.pop()

    print(f"\nProduto removido: {removido}")
    print(f"Preço removido: R$ {preco_removido:.2f}")
else:
    print("Estoque vazio!")
```
